### First Create Catalog

```python
catalog_name = 'DataModeling'
volume_name = 'data_source'
```

```python
%sql
drop table datamodeling.default.source_table
```

```python
# Create catalog, schema and volume if they donot exist
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog_name}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog_name}.default.{volume_name}")
```

```python
%sql
create table if not exists datamodeling.default.source_table
(
    order_id int primary key,
    order_date date,
    customer_id int,
    customer_name varchar(100),
    customer_email varchar(100),
    product_id int,
    product_name varchar(100),
    product_category varchar(100),
    product_price decimal(10,2),
    quantity int,
    revenue decimal(10,2),
    country varchar(100),
    last_updated date
)
```
